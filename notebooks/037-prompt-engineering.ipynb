## 1. Specify task and boundaries

A useful prompt makes the task, permitted evidence, audience, and constraints explicit.


````python
question = "Can I return an item after 20 days?"
policy = "Returns are accepted within 30 days with a receipt."
instruction = "Answer using only the supplied policy; say when the evidence is insufficient."
````

## 2. Request a structured response

A requested schema makes downstream validation easier, but the application must still validate the model response.


````python
prompt = (
    f"{instruction}\nPolicy: {policy}\nQuestion: {question}\n"
    "Return JSON with keys answer and supporting_quote."
)
print(prompt)
````

## 3. Test prompt behavior with examples

A prompt should be evaluated on representative, ambiguous, and adversarial questions rather than one hand-picked success.


````python
test_questions = [
    "Can I return this after 20 days?",
    "Can I return it after 50 days?",
    "Can I return it without a receipt?",
]
print("evaluation cases:", len(test_questions))
````
