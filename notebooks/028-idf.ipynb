# Inverse document frequency (IDF): a hands-on implementation

**Track:** Search · **Guide:** [`028-idf.md`](../guides/concepts/028-idf.md)

Inverse document frequency reduces the weight of terms that appear in many documents and increases the weight of rarer terms.

> **Intuition:** A detail that distinguishes one folder matters more than a label printed on every folder.

## 1. Build document frequencies

Document frequency counts how many documents contain a term, not how many times it appears overall.


````python
import math
import re
from collections import Counter

documents = [
    "reset password reset account",
    "account billing question",
    "password reset failed again",
    "account access help",
]
token_sets = [set(re.findall(r"[a-z]+", doc.lower())) for doc in documents]
document_frequency = Counter(term for terms in token_sets for term in terms)
print("document count:", len(documents))
print("document frequencies:", document_frequency)
````

## 2. Calculate smoothed IDF

Smoothing avoids division by zero and keeps IDF finite; frequent terms receive lower weights.


````python
N = len(documents)
idf = {
    term: math.log((1 + N) / (1 + frequency)) + 1
    for term, frequency in document_frequency.items()
}
for term in ("account", "reset", "billing"):
    print(f"{term:8s} df={document_frequency[term]} idf={idf[term]:.3f}")
````

## 3. Explain a query-weighting consequence

A rare billing term is more discriminative in this corpus than a common account term; this depends on the collection.


````python
query = ["account", "billing"]
query_weights = {term: idf[term] for term in query}
print("query weights:", query_weights)
assert query_weights["billing"] > query_weights["account"]
````
