## 1. Convert logits to probabilities

Temperature rescales logits before softmax; lower values sharpen the distribution and higher values flatten it.


````python
import numpy as np

logits = np.array([2.0, 1.0, 0.1])
````

## 2. Compare temperatures

Sampling settings affect variety and repeatability; they do not repair an incorrect or unsupported answer.


````python
def softmax(values):
    shifted = values - values.max()
    probabilities = np.exp(shifted)
    return probabilities / probabilities.sum()

for temperature in (0.5, 1.0, 1.8):
    print(temperature, softmax(logits / temperature).round(3).tolist())
````

## 3. Sample reproducibly

A fixed random seed makes this demonstration repeatable; production generation also needs token and time limits.


````python
rng = np.random.default_rng(2)
probabilities = softmax(logits / 0.8)
tokens = ["A", "B", "C"]
print([tokens[index] for index in rng.choice(len(tokens), size=5, p=probabilities)])
````
