## 1. Represent queries, keys, and values

A query is compared with keys; the resulting weights mix value vectors.


````python
import numpy as np

query = np.array([1.0, 0.0])
keys = np.array([[1.0, 0.0], [0.0, 1.0]])
values = np.array([[3.0, 0.0], [0.0, 6.0]])
````

## 2. Compute scaled attention weights

Scaling dot products by the square root of the key dimension helps control score magnitude.


````python
scores = query @ keys.T / np.sqrt(keys.shape[1])
weights = np.exp(scores - scores.max())
weights /= weights.sum()
print("scores:", scores.round(3), "weights:", weights.round(3))
````

## 3. Combine values

The output is a weighted sum of values; multi-head attention repeats this operation with learned projections.


````python
attended = weights @ values
print("attended representation:", attended.round(3))
````
