## 1. Inspect raw text before normalization

Raw text can contain negation, identifiers, punctuation, and casing that affect the task.


````python
import re

text = "Can't reset device E-1042!"
print("raw:", text)
````

## 2. Tokenize with explicit rules

This simple tokenizer retains contractions and alphanumeric product IDs; production tokenizers are language- and model-specific.


````python
normalized = text.lower()
tokens = re.findall(r"[a-z]+-\d+|[a-z]+(?:'[a-z]+)?|\d+|[^\w\s]", normalized)
print("tokens:", tokens)
````

## 3. Avoid harmful normalization

Removing punctuation, negation, or identifiers can change meaning; compare preprocessing choices on a held-out task set.


````python
assert "can't" in tokens
assert "e-1042" in tokens
print("important distinctions preserved")
````
