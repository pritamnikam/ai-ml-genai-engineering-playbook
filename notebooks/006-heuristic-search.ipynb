## 1. Define a weighted route-planning problem

A* searches paths using cost so far plus a heuristic estimate to the goal.


````python
import heapq

graph = {
    "A": {"B": 2, "C": 4},
    "B": {"C": 1, "D": 5},
    "C": {"D": 2},
    "D": {},
}
heuristic = {"A": 4, "B": 3, "C": 2, "D": 0}
````

## 2. Implement A* with path reconstruction

A priority queue expands the most promising node. The example heuristic never overestimates the remaining path cost.


````python
def a_star(graph, start, goal, heuristic):
    frontier = [(heuristic[start], 0, start)]
    best_cost = {start: 0}
    parent = {}
    while frontier:
        _estimated_total, cost, node = heapq.heappop(frontier)
        if cost != best_cost[node]:
            continue
        if node == goal:
            path = [node]
            while node in parent:
                node = parent[node]
                path.append(node)
            return list(reversed(path)), cost
        for neighbor, edge_cost in graph[node].items():
            candidate = cost + edge_cost
            if candidate < best_cost.get(neighbor, float("inf")):
                best_cost[neighbor] = candidate
                parent[neighbor] = node
                heapq.heappush(
                    frontier, (candidate + heuristic[neighbor], candidate, neighbor)
                )
    return None, float("inf")
````

## 3. Check the route and optimal cost

With a nonnegative admissible heuristic, A* finds an optimal route. Changing the heuristic can change efficiency and correctness.


````python
path, cost = a_star(graph, "A", "D", heuristic)
print("route:", path, "cost:", cost)
assert path == ["A", "B", "C", "D"] and cost == 5
````
