## 1. Build an interaction matrix

Rows are users, columns are items, and missing values mean no observed interaction—not necessarily dislike.


````python
import numpy as np

users = ["A", "B", "C", "D"]
items = ["book-1", "book-2", "book-3", "book-4"]
ratings = np.array([
    [5, 4, 0, 0], [5, 0, 4, 0], [0, 5, 4, 0], [0, 0, 1, 5]
], dtype=float)
print("user-item matrix shape:", ratings.shape)
````

## 2. Score items by similar users

A simple neighborhood score uses similarity-weighted interactions and excludes already-seen items.


````python
target_user = 0
similarities = ratings @ ratings[target_user] / (
    np.linalg.norm(ratings, axis=1) * np.linalg.norm(ratings[target_user]) + 1e-9
)
candidate_scores = similarities @ ratings / (np.abs(similarities).sum() + 1e-9)
candidate_scores[ratings[target_user] > 0] = -np.inf
print("candidate scores:", candidate_scores.round(2).tolist())
````

## 3. Rank recommendations and state limitations

Offline metrics should use a time-based holdout; consider cold-start users, popularity bias, privacy, and catalog coverage.


````python
recommendations = sorted(
    zip(items, candidate_scores), key=lambda pair: pair[1], reverse=True
)
print("top recommendations:", recommendations[:2])
````
