## 2. Monitor input drift

A change in feature distribution may indicate a new operating regime or a broken data pipeline; it is not itself proof of lower quality.


````python
training = [10, 11, 9, 10, 10]
serving = [18, 19, 17, 18, 20]
mean_shift = sum(serving) / len(serving) - sum(training) / len(training)
print("mean shift:", mean_shift)
````

## 3. Define alert and rollback behavior

Monitoring needs an owner and an action; retain the previous known-good artifact and verify rollback procedures.


````python
alert = abs(mean_shift) > 3
action = "investigate and consider rollback" if alert else "continue monitoring"
print({"alert": alert, "action": action, "rollback_version": "ticket-risk-v0"})
````
