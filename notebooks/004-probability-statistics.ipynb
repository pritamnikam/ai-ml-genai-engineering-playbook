## 1. Summarize observations

A sample mean and event rate summarize observed data but do not remove sampling uncertainty.


````python
import numpy as np

readings = np.array([18.2, 19.1, 17.7, 20.0, 18.9])
print("mean:", round(float(readings.mean()), 2))
print("sample standard deviation:", round(float(readings.std(ddof=1)), 2))
````

## 2. Estimate a probability

An empirical probability is a frequency estimate; the estimate is less certain with few observations.


````python
failures, trials = 3, 20
estimated_rate = failures / trials
print(f"estimated failure rate: {estimated_rate:.1%} from {trials} trials")
````

## 3. Apply Bayes' rule

Bayes' rule combines a prior event rate with evidence sensitivity and false-positive rate.


````python
prior = 0.02
sensitivity = 0.90
false_positive_rate = 0.05
posterior = sensitivity * prior / (
    sensitivity * prior + false_positive_rate * (1 - prior)
)
print(f"probability of failure after a positive alert: {posterior:.1%}")
````
