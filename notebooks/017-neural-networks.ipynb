## 1. Create a tiny feed-forward network

A layer applies a weighted sum and a nonlinear activation to its inputs.


````python
import numpy as np

sigmoid = lambda value: 1 / (1 + np.exp(-value))
x = np.array([0.6, 0.2])
W1 = np.array([[0.5, -0.4], [0.3, 0.8]])
b1 = np.array([0.0, 0.1])
W2 = np.array([[0.7], [-0.2]])
b2 = np.array([0.0])
````

## 2. Forward-propagate through layers

The hidden representation is transformed into an output probability; training adjusts weights to reduce a loss.


````python
hidden = np.tanh(x @ W1 + b1)
probability = sigmoid(hidden @ W2 + b2)
print("hidden:", hidden.round(3), "probability:", probability.round(3))
````

## 3. Check how a parameter affects the output

A small perturbation illustrates sensitivity; backpropagation computes these derivatives efficiently for all parameters.


````python
perturbed_W2 = W2.copy()
perturbed_W2[0, 0] += 0.01
perturbed_probability = sigmoid(hidden @ perturbed_W2 + b2)
print("output change:", round(float((perturbed_probability - probability).item()), 6))
````
