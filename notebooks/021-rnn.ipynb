## 1. Encode a sequence

A recurrent network consumes one timestep at a time. This toy sequence contains a persistent signal followed by a change.


````python
import numpy as np

sequence = np.array([0.9, 0.8, 0.7, 0.1, 0.0], dtype=float)
print("timesteps:", len(sequence), "values:", sequence)
````

## 2. Run a simple recurrent cell

At each step, the hidden state mixes the new input with the previous state. Parameters are fixed here so the recurrence is easy to inspect.


````python
def run_rnn(values, input_weight=0.8, recurrent_weight=0.5):
    state = 0.0
    states = []
    for value in values:
        state = np.tanh(input_weight * value + recurrent_weight * state)
        states.append(state)
    return np.array(states)

hidden_states = run_rnn(sequence)
print("hidden states:", np.round(hidden_states, 3))
````

## 3. Use the final state as a summary

A real RNN learns its weights from examples. Here, threshold the final state and inspect how a sequential summary differs from classifying each input alone.


````python
threshold = 0.25
summary_prediction = "signal" if hidden_states[-1] > threshold else "no-signal"
print("last raw value:", sequence[-1])
print("last hidden state:", round(float(hidden_states[-1]), 3))
print("sequence-level prediction:", summary_prediction)
````
