## 1. Compare model families

VAEs learn a probabilistic latent representation, GANs train a generator against a discriminator, and diffusion models learn denoising steps.


````python
model_families = {
    "VAE": "encode to a regularized latent distribution, then decode",
    "GAN": "train a generator and discriminator in competition",
    "diffusion": "learn to reverse a gradual noising process",
}
for family, idea in model_families.items():
    print(f"{family}: {idea}")
````

## 2. Simulate a diffusion forward process

A forward process progressively adds noise; training learns a denoiser for reversing this process.


````python
import numpy as np

rng = np.random.default_rng(7)
clean = np.array([0.2, 0.8, 0.4, 0.6])
noise = rng.normal(size=clean.shape)
alpha = 0.7
noisy = np.sqrt(alpha) * clean + np.sqrt(1 - alpha) * noise
print("noisy sample:", noisy.round(3))
````

## 3. Inspect a known-noise reconstruction

This oracle reconstruction knows the injected noise exactly; a real generative model must estimate it from data and never gets this shortcut.


````python
reconstructed = (noisy - np.sqrt(1 - alpha) * noise) / np.sqrt(alpha)
print("reconstruction:", reconstructed.round(3))
assert np.allclose(reconstructed, clean)
````
