# Term frequency (TF): a hands-on implementation

Term frequency measures how often a term occurs in a document, sometimes normalized by document length.

> **Intuition:** Count how many times a keyword appears in one report before comparing it with other reports.

## 1. Tokenize a document

Term frequency starts with counting tokens in each document; this lesson uses a tiny support-ticket corpus.


````python
import re
from collections import Counter

documents = [
    "reset password reset account",
    "account billing question",
    "password reset failed again",
]
tokenized = [re.findall(r"[a-z]+", doc.lower()) for doc in documents]
counts = [Counter(tokens) for tokens in tokenized]
print("first document counts:", counts[0])
````

## 2. Compare raw and normalized term frequency

Raw count grows with document length. Dividing by length gives a simple length-normalized alternative.


````python
term = "reset"
for index, (tokens, counter) in enumerate(zip(tokenized, counts), start=1):
    raw = counter[term]
    normalized = raw / len(tokens)
    print(f"doc {index}: count={raw}, normalized_tf={normalized:.3f}")
````

## 3. Inspect the document-term matrix

A fixed vocabulary maps each token to a consistent column so term counts can be compared across documents.


````python
vocabulary = sorted({token for tokens in tokenized for token in tokens})
term_matrix = [
    [counter[token] for token in vocabulary]
    for counter in counts
]
print("columns:", vocabulary)
for row in term_matrix:
    print(row)
````
