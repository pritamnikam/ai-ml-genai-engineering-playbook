## 1. Prepare a nonlinear classification task

A multilayer perceptron is a small example of a deep model; its hidden layers learn intermediate representations.


````python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split

X, y = make_moons(n_samples=240, noise=0.18, random_state=8)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=8, stratify=y
)
````

## 2. Train a multilayer network

Deep models are not automatically better; compare against simpler baselines and tune on validation data.


````python
from sklearn.neural_network import MLPClassifier

network = MLPClassifier(
    hidden_layer_sizes=(16, 8), max_iter=500, random_state=8, early_stopping=True
).fit(X_train, y_train)
````

## 3. Evaluate the held-out set

Report held-out performance and resource use; this small CPU example is not evidence that a deep model is needed.


````python
print("held-out accuracy:", round(network.score(X_test, y_test), 3))
print("training iterations:", network.n_iter_)
````
