Tool calling lets a model request a named operation with structured arguments; application code validates and authorizes before execution.

> **Intuition:** The model may fill in a request form, but a trusted operator checks the form and permissions before acting.

## 1. Define a narrow tool contract

Tools should expose the smallest capability needed, with typed arguments and bounded effects.


````python
ALLOWED_TOOLS = {"lookup_order"}
orders = {"ORD-1042": {"status": "shipped"}}
````

## 2. Validate and authorize before dispatch

Never execute arbitrary model-generated code or trust arguments without validation and authorization.


````python
def dispatch_tool(name, arguments):
    if name not in ALLOWED_TOOLS:
        raise ValueError(f"tool not allowed: {name}")
    order_id = arguments.get("order_id")
    if not isinstance(order_id, str) or order_id not in orders:
        raise ValueError("unknown or invalid order_id")
    return orders[order_id]

print(dispatch_tool("lookup_order", {"order_id": "ORD-1042"}))
````

## 3. Require approval for side effects

Read-only tools can often be automatic; sending messages, changing records, or spending money needs explicit policy and often human confirmation.


````python
def approve_action(action, approved):
    if action in {"send_email", "issue_refund"} and not approved:
        return "awaiting_human_approval"
    return "ready_to_execute"

print(approve_action("issue_refund", approved=False))
````
