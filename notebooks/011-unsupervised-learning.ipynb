## 1. Create unlabeled data

No target labels are supplied. Standardization can matter because distance-based methods are scale-sensitive.


````python
import numpy as np
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

X, hidden_groups = make_blobs(
    n_samples=120, centers=3, cluster_std=1.1, random_state=5
)
X_scaled = StandardScaler().fit_transform(X)
print("observed feature shape:", X_scaled.shape)
````

## 2. Discover groups

K-means assigns points to nearby centroids; the chosen number of clusters is an assumption, not discovered truth.


````python
from sklearn.cluster import KMeans

clusterer = KMeans(n_clusters=3, n_init=10, random_state=5).fit(X_scaled)
cluster_ids = clusterer.labels_
print("cluster sizes:", np.bincount(cluster_ids).tolist())
````

## 3. Inspect and qualify the result

Compare clusters with domain knowledge and stability checks; do not treat synthetic hidden labels as available in real unsupervised work.


````python
for cluster_id in range(3):
    center = X[cluster_ids == cluster_id].mean(axis=0)
    print(cluster_id, "original-unit center:", center.round(2).tolist())
````

## 4. Reduce dimensionality with PCA

Principal component analysis projects data onto directions that preserve variance; it is useful for compression and visualization, not automatically for better predictions.


````python
from sklearn.decomposition import PCA

pca = PCA(n_components=2).fit(X_scaled)
coordinates = pca.transform(X_scaled)
print("explained variance:", pca.explained_variance_ratio_.round(3).tolist())
print("projected shape:", coordinates.shape)
````
