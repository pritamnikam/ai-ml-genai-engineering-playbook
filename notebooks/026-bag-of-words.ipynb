# Bag-of-words representations: a hands-on implementation

Bag of words represents each document by token counts, ignoring word order while retaining vocabulary membership and frequency.

> **Intuition:** It is a shopping receipt of words: what appeared and how often, but not the order they were used.

## 1. Build a vocabulary

A fixed vocabulary maps each token to a stable column in the document-term matrix.


````python
from collections import Counter

documents = ["reset password", "password reset failed", "billing question"]
tokenized = [document.lower().split() for document in documents]
vocabulary = sorted({token for document in tokenized for token in document})
term_index = {term: index for index, term in enumerate(vocabulary)}
print(term_index)
````

## 2. Count terms per document

Bag of words stores counts and deliberately ignores token order.


````python
import numpy as np

matrix = np.zeros((len(documents), len(vocabulary)), dtype=int)
for row, tokens in enumerate(tokenized):
    counts = Counter(tokens)
    for term, count in counts.items():
        matrix[row, term_index[term]] = count
print("columns:", vocabulary)
print(matrix)
````

## 3. Inspect what order was lost

Documents with the same word counts have the same bag-of-words representation even if their word order differs.


````python
first, second = "dog bites person", "person bites dog"
def counts(text):
    return Counter(text.split())
print("same counts:", counts(first) == counts(second))
````
