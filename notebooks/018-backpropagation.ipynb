# Backpropagation and gradient descent: a hands-on implementation

Backpropagation applies the chain rule to compute how each parameter affects loss; gradient descent updates parameters to reduce that loss.

> **Intuition:** Trace an error backward through a chain of calculations, then adjust each knob in the direction that reduces it.

## 1. Define a differentiable prediction and loss

For a one-weight linear model, squared-error loss has a simple derivative that can be derived by the chain rule.


````python
x, target, weight, bias = 2.0, 5.0, 0.5, 0.0
prediction = weight * x + bias
loss = (prediction - target) ** 2
print({"prediction": prediction, "loss": loss})
````

## 2. Compute gradients

The derivative of squared error with respect to the weight includes the input factor from the prediction equation.


````python
error = prediction - target
gradient_weight = 2 * error * x
gradient_bias = 2 * error
print({"d_loss_d_weight": gradient_weight, "d_loss_d_bias": gradient_bias})
````

## 3. Update parameters and measure loss again

Gradient descent moves parameters opposite the gradient; a learning rate that is too large can make loss diverge.


````python
learning_rate = 0.1
weight -= learning_rate * gradient_weight
bias -= learning_rate * gradient_bias
new_loss = (weight * x + bias - target) ** 2
print({"updated_parameters": (weight, bias), "loss_after_update": new_loss})
assert new_loss < loss
````
