## 1. Represent items as vectors

Real text embeddings are learned dense vectors. For a transparent offline example, define small vectors that encode a toy set of product concepts.


````python
import numpy as np

items = {
    "car": np.array([0.95, 0.05, 0.0]),
    "automobile": np.array([0.90, 0.10, 0.0]),
    "truck": np.array([0.82, 0.18, 0.0]),
    "pasta recipe": np.array([0.0, 0.05, 0.95]),
}
````

## 2. Compare with cosine similarity

Cosine similarity compares direction and is commonly used for normalized embeddings.


````python
def cosine_similarity(left, right):
    denominator = np.linalg.norm(left) * np.linalg.norm(right)
    if denominator == 0:
        raise ValueError("cosine similarity is undefined for a zero vector")
    return float(np.dot(left, right) / denominator)

query = items["automobile"]
scores = {name: cosine_similarity(query, vector) for name, vector in items.items()}
print(sorted(scores.items(), key=lambda pair: -pair[1]))
````

## 3. Check vector contracts

A production embedding index requires a consistent model, vector dimension, normalization convention, and version across queries and documents.


````python
dimensions = {vector.shape for vector in items.values()}
print("vector shapes:", dimensions)
assert dimensions == {(3,)}
print("These hand-authored vectors are illustrative, not semantic model embeddings.")
````
