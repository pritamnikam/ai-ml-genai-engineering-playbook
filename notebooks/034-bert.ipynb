## 1. Represent a masked-token prediction task

BERT's masked-language modeling objective predicts a hidden token using context on both its left and right.


````python
import re
from collections import Counter, defaultdict

sentences = [
    "the battery lasts all day",
    "the battery charges quickly",
    "the screen looks bright",
    "the screen stays bright",
    "the battery feels warm",
]
tokens = [re.findall(r"[a-z]+", sentence) for sentence in sentences]
print("sample:", tokens[0])
````

## 2. Learn a toy bidirectional context table

For each center word, count neighboring words on both sides. This is a transparent miniature, not a Transformer or pretrained BERT.


````python
context_counts = defaultdict(Counter)
for sentence in tokens:
    for index in range(1, len(sentence) - 1):
        context = (sentence[index - 1], sentence[index + 1])
        context_counts[context][sentence[index]] += 1

context = ("the", "lasts")
print("context:", context, "candidates:", context_counts[context])
````

## 3. Predict a masked token from both sides

Score vocabulary candidates with their observed left/right context pair and report when the training corpus has no matching evidence.


````python
def predict_mask(left, right):
    candidates = context_counts[(left, right)]
    if not candidates:
        return []
    return candidates.most_common()

print("[the] battery [lasts] ->", predict_mask("the", "lasts"))
print("[the] screen [charges] ->", predict_mask("the", "charges"))
````
