## 1. Create a reproducible dataset

Keep the data split identical so model comparisons reflect modeling choices rather than different examples.


````python
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split

X, y = make_moons(n_samples=300, noise=0.2, random_state=12)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, stratify=y, random_state=12
)
````

## 2. Compare a linear model and a tree

A linear decision boundary is simple; a tree can model nonlinear partitions but may overfit.


````python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

models = {
    "logistic": LogisticRegression(),
    "shallow_tree": DecisionTreeClassifier(max_depth=3, random_state=12),
}
for model in models.values():
    model.fit(X_train, y_train)
````

## 3. Measure on held-out data

Compare accuracy and complexity, and then inspect errors and robustness before choosing a model.


````python
for name, model in models.items():
    print(name, "accuracy:", round(model.score(X_test, y_test), 3))
````
