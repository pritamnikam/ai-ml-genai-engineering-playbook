# TF-IDF: a hands-on implementation

TF-IDF weights a term by its frequency in one document and discounts terms common across the collection.

> **Intuition:** A word earns attention when it is repeated in this report but is uncommon in the whole library.

This lesson builds a small working example in stages: prepare inputs, implement

## 1. Construct a tiny document-term matrix

TF-IDF combines within-document term frequency with collection-wide inverse document frequency.


````python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "reset password reset account",
    "account billing address",
    "password reset failed",
]
vectorizer = TfidfVectorizer(norm=None, smooth_idf=True)
matrix = vectorizer.fit_transform(documents)
terms = vectorizer.get_feature_names_out()
print("terms:", list(terms))
print("matrix shape:", matrix.shape)
````

## 2. Inspect TF-IDF weights

Common terms are discounted relative to terms concentrated in fewer documents. Exact values depend on the vectorizer's TF and smoothing conventions.


````python
for document_index, row in enumerate(matrix.toarray(), start=1):
    weighted_terms = {
        term: round(float(weight), 3)
        for term, weight in zip(terms, row)
        if weight
    }
    print(f"document {document_index}:", weighted_terms)
````

## 3. Rank documents for a query

Cosine similarity compares query and document vectors; normalization makes it less sensitive to document length.


````python
from sklearn.metrics.pairwise import cosine_similarity

query_vector = vectorizer.transform(["password reset"])
scores = cosine_similarity(query_vector, matrix).ravel()
ranking = scores.argsort()[::-1]
for index in ranking:
    print(f"score={scores[index]:.3f}  {documents[index]}")
````
